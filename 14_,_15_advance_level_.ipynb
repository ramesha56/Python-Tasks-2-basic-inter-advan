```python
# Task 14: Restaurant Ordering System

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

print("Restaurant Menu")
print("Burger - Rs.500")
print("Pizza - Rs.1200")
print("Pasta - Rs.800")
print("Drink - Rs.200")

orders = []

while True:

    item = input("\nEnter item name: ").title()

    if item in menu:
        quantity = int(input("Enter quantity: "))

        subtotal = menu[item] * quantity

        order = {
            "item": item,
            "quantity": quantity,
            "subtotal": subtotal
        }

        orders.append(order)

        print("Item added successfully.")

    else:
        print("Item is not available.")

    more = input("Do you want to order more? ").lower()

    if more == "no":
        break


def receipt():
    total = 0

    print("\n----- RECEIPT -----")

    for order in orders:
        print(order["item"], "x", order["quantity"], "=", order["subtotal"])
        total = total + order["subtotal"]

    print("Total:", total)

    if total >= 3000:
        discount = total * 0.10
    else:
        discount = 0

    tax = total * 0.13

    final_bill = total - discount + tax

    print("Discount:", discount)
    print("Tax:", tax)
    print("Final Bill:", final_bill)


receipt()
```

```python
# Task 15: Clinic Appointment Manager

patients = []


def add_appointment():
    patient_id = input("Enter Patient ID: ")


    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists.")
            return

    name = input("Enter Patient Name: ").title()
    age = int(input("Enter Age: "))
    problem = input("Enter Disease/Problem: ").title()
    time = input("Enter Appointment Time: ")
    emergency = input("Emergency patient? yes/no: ").lower()

    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "problem": problem,
        "time": time,
        "emergency": emergency
    }

    patients.append(patient)

    print("Appointment added successfully.")


def search_patient():
    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient Name:", patient["name"])
            print("Age:", patient["age"])
            print("Problem:", patient["problem"])
            print("Appointment Time:", patient["time"])
            print("Emergency:", patient["emergency"])
            return

    print("Patient not found.")


def display_appointments():
    print("\n--- Appointments ---")


    for patient in patients:
        if patient["emergency"] == "yes":
            print(patient["id"], "-", patient["name"], "-", "Emergency")


    for patient in patients:
        if patient["emergency"] == "no":
            print(patient["id"], "-", patient["name"], "-", "Normal")


def cancel_appointment():
    patient_id = input("Enter Patient ID to cancel: ")

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Appointment cancelled.")
            return

    print("Patient not found.")


def clinic_summary():
    normal = 0
    emergency = 0

    for patient in patients:
        if patient["emergency"] == "yes":
            emergency = emergency + 1
        else:
            normal = normal + 1

    print("\n--- Clinic Summary ---")
    print("Normal Patients:", normal)
    print("Emergency Patients:", emergency)
    print("Total Patients:", normal + emergency)


while True:

    print("\n--- Clinic Appointment Manager ---")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_appointment()

    elif choice == "2":
        search_patient()

    elif choice == "3":
        display_appointments()

    elif choice == "4":
        cancel_appointment()

    elif choice == "5":
        clinic_summary()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
