```python
#tast 1

name = input("Enter your name: ")
age = int(input("Enter your age: "))

if age < 5 or age > 100:
    price = 0
elif age <= 12:
    price = 300
elif age <= 59:
    price = 700
else:
    price = 400

print("\nCustomer:", name)
print("Ticket Price: Rs.", price)
```

```python
##task 2
celsius = int(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

if celsius < 10:
    weather = "Very Cold"
elif celsius <= 19:
    weather = "Cold"
elif celsius <= 29:
    weather = "Normal"
else:
    weather = "Hot"

print("Fahrenheit:", fahrenheit)
print("Weather:", weather)
```

```python
#task 4


total_steps = 0

for day in range(7):
    steps = int(input("Enter steps: "))
    total_steps = total_steps + steps

average = total_steps / 7

print("Total Steps:", total_steps)
print("Daily Average:", average)

if average >= 5000:
    print("Status: Active Week")
else:
    print("Status: Needs More Activity")
```

```python

##task 5

secret_pin = "1234"

for attempt in range(3):
    pin = input("Enter PIN: ")

    if pin == secret_pin:
        print("Access Granted")
        break
    else:
        print("Incorrect PIN")

        if attempt == 2:
            print("Account Blocked")
        else:
            print("Remaining attempts:", 2 - attempt)
```
